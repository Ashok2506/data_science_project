# How does the price vary

```python
final_data
```

```python
# extracting respected hotel data from final_data to find price variation 
resort_data = final_data[(final_data["hotel"]== "Resort Hotel") & (final_data["is_canceled"]==0)]

city_data = final_data[(final_data["hotel"]== "City Hotel") & (final_data["is_canceled"]==0)]
```

```python
#how many people came to the resort hotel per month

resort_hotel = resort_data.groupby(["arrival_date_month"])["adr"].mean().reset_index()
print(resort_hotel)
```

```python
#how many people came to the city hotel per month

city_hotel = city_data.groupby(["arrival_date_month"])["adr"].mean().reset_index()
print(city_hotel)
```

```python
#merging two hotels to find the price variations
final = resort_hotel.merge(city_hotel, on = "arrival_date_month")
final.columns = ["Month","Resort_hotel_price","City_hotel_price"]
```

```python
print(final)
```

# sorting the month

```python
from calendar import month_name

def sort_month(df,colname):
    month_dict = { j:i for i,j in enumerate(month_name)}
    df["month_num"] = df[colname].apply(lambda x :month_dict[x])
    return df.sort_values(by="month_num").reset_index().drop(["month_num","index"],axis = 1)
```

```python
sort_month(final,"Month")
```

```python
final.plot(kind = "line",x="Month",y=["Resort_hotel_price","City_hotel_price"])
plt.show()
```

# select numeric columns

```python
col_list = ["previous_bookings_not_canceled","days_in_waiting_list","stays_in_weekend_nights","children"]
#using list comprehension to find those columns which we need to train model 

num_features = [col for col in final_data.columns if final_data[col].dtype != "O" and col not in col_list]
print(num_features)
```

```python
num_data = final_data[num_features]
```

```python
num_data.head()
```

```python
num_data.shape
```

# select categorical columns

```python
cat_not = ["reservation_status","booking_changes","country"]

cat_features = [col for col in final_data.columns if final_data[col].dtype == "O" and col not in cat_not]
print(cat_features)
```

# convert date column  object to datetime format

```python
cat_data = final_data[cat_features]
cat_data.head()
```

```python
import warnings
from warnings import filterwarnings
filterwarnings("ignore")

cat_data["reservation_status_date"] = pd.to_datetime(cat_data["reservation_status_date"])
cat_data["reservation_status_date"].dtype
```

```python
cat_data["year"] = cat_data["reservation_status_date"].dt.year
cat_data["month"] = cat_data["reservation_status_date"].dt.month
cat_data["day"] = cat_data["reservation_status_date"].dt.day
cat_data.head()
```

```python
cat_data.drop("reservation_status_date", axis = 1 , inplace = True)
cat_data.head()
```

```python
cat_data.shape
```

# Feature encoding

```python
cat_data["cancellation"] = final_data["is_canceled"]
```

```python
cat_data.head()
```

```python
def mean_encode(df,col,mean_col):
    df_dict = df.groupby([col])[mean_col].mean().to_dict()
    df[col] = df[col].map(df_dict)
    return df

for col in cat_data.columns[0:9]:
    data_cat = mean_encode(cat_data,col,"cancellation")
    
```

```python
data_cat.head()
```

```python
data_cat = data_cat.drop("cancellation",axis = 1)
print(data_cat.head())
```

# prepare data

```python
categorical_data = data_cat

dataframe = pd.concat([num_data,categorical_data],axis = 1)
```

```python
dataframe.head()
```

# handling outliers

```python
sns.distplot(dataframe['lead_time'])
```

```python
def handle_outliers(col):
    dataframe[col] = np.log1p(dataframe[col])
    
```

```python
handle_outliers('lead_time')
```

```python
handle_outliers('adr')
```

```python
sns.distplot(dataframe['adr'])
```

```python
dataframe.isnull().sum()
```

```python
dataframe.dropna(inplace = True)
```

# feature importance

```python
# select dependent and independent variables

y = dataframe["is_canceled"] #dependent variable

x = dataframe.drop("is_canceled",axis = 1) # independent variable

```

```python
from sklearn.linear_model import Lasso
from sklearn.feature_selection import SelectFromModel

feature_sel_model = SelectFromModel(Lasso(alpha = 0.005,random_state = 0))
feature_sel_model.fit(x,y)

feature_sel_model.get_support()
```

```python
cols = x.columns
```

```python
cols
```

```python
selected_feature = cols[(feature_sel_model.get_support())]
```

```python
print(selected_feature)
```

```python
print(f"Total features {x.shape[1]}")
```

```python
print(f"selected features {len(selected_feature)}")
```

```python
x = x[selected_feature]
x.head()
```

```python
y.head()
```

# split the data and build the model

```python
# x -- independent variables
# y -- dependent variables

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y, train_size = 0.75,random_state = 45)
```

# implement logistic regression

```python
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression()
```

```python
logistic_model.fit(x_train,y_train)

y_pred = logistic_model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_pred,y_test))
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_pred,y_test))
```

# implementing different classification algorithms

### logistic regression
### Naive Bayes
### Random Forest
### Decision Tree
### KNN

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = []
models.append(("Naive Bayes", GaussianNB()))
models.append(("Random Forest",RandomForestClassifier()))
models.append(("Decision Tree",DecisionTreeClassifier()))
models.append(("KNN",KNeighborsClassifier()))
```

```python
for name, model in models:
    print (name)
    model.fit(x_train,y_train)

    # make a prediction
    prediction = model.predict(x_test)

    #evaluate the model
    from sklearn.metrics import confusion_matrix
    print(confusion_matrix(prediction,y_test))

    from sklearn.metrics import accuracy_score
    print(accuracy_score(prediction,y_test))

    print("\n")
```
